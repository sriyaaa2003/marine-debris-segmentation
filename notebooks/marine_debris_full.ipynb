```python
# ============================================================
# Band and Class Mapping
# Purpose: Define Sentinel-2 band mappings and semantic class mappings
# ============================================================

B01, B02, B03, B04 = 0, 1, 2, 3
B05, B06, B07, B08 = 4, 5, 6, 7
B8A, B11, B12      = 8, 9, 10

BAND_NAMES = ['B01','B02','B03','B04','B05',
              'B06','B07','B08','B8A','B11','B12']

CLASS_MERGE = {
    0:  -1,
    1:   0,   # Marine Debris
    2:   1,   # Dense Sargassum   Sargassum
    3:   1,   # Sparse Sargassum  Sargassum
    4:   2,   # Natural Organic Material
    5:   3,   # Ship
    6:   4,   # Clouds
    7:   5,   # Marine Water      Water
    8:   5,   # Sediment-Laden    Water
    9:   6,   # Foam
    10:  5,   # Turbid Water      Water
    11:  5,   # Shallow Water     Water
    12:  7,   # Waves
    13:  8,   # Cloud Shadows
    14:  9,   # Wakes
    15:  5,   # Mixed Water       Water
}

NUM_CLASSES = 10

CLASS_NAMES = [
    'Marine Debris',
    'Sargassum',
    'Natural Organic Mat.',
    'Ship',
    'Clouds',
    'Water',
    'Foam',
    'Waves',
    'Cloud Shadows',
    'Wakes',
]

CONF_WEIGHTS = {0: 0.0, 1: 0.5, 2: 0.75, 3: 1.0}

WL_RED   = 665.0
WL_RE2   = 740.0
WL_NIR   = 842.0
WL_SWIR1 = 1614.0

MERGE_LUT = np.array([CLASS_MERGE[k] for k in range(16)], dtype=np.int64)
CONF_LUT  = np.array([CONF_WEIGHTS[k] for k in range(4)],  dtype=np.float32)

print(f"NUM_CLASSES : {NUM_CLASSES}")
print(f"CLASS_NAMES : {CLASS_NAMES}")
print("Constants OK")
```

```python
# ============================================================
# Patch Lookup Table
# Purpose: Build lookup table for image, label, and confidence patch files
# ============================================================
def build_patch_lookup(patches_dir: Path) -> dict:
    lookup = {}
    for tif in sorted(patches_dir.rglob('*.tif')):
        name = tif.stem
        if name.endswith('_cl'):
            pid = name[:-3]
            lookup.setdefault(pid, {})['label'] = tif
        elif name.endswith('_conf'):
            pid = name[:-5]
            lookup.setdefault(pid, {})['conf'] = tif
        else:
            lookup.setdefault(name, {})['image'] = tif
    complete = {pid: v for pid, v in lookup.items()
                if 'image' in v and 'label' in v}
    print(f'Patches found: {len(complete)}')
    return complete

PATCH_LOOKUP = build_patch_lookup(PATCHES_DIR)
print('Patch lookup OK')
```

```python
# ============================================================
# Spectral Feature Engineering
# Purpose: Compute engineered multispectral indices for marine debris detection
# ============================================================

def compute_spectral_indices(image: np.ndarray) -> np.ndarray:
    """
    Compute FDI, NDVI, NDWI, FAI from raw un-normalised reflectance.
    Args:
        image: (11, H, W) float32
    Returns:
        (4, H, W) float32
    """
    eps   = 1e-8
    nir   = image[B08].astype(np.float32)
    red   = image[B04].astype(np.float32)
    green = image[B03].astype(np.float32)
    re2   = image[B06].astype(np.float32)
    swir1 = image[B11].astype(np.float32)

    # FDI — Biermann et al. 2020
    nir_bl_fdi = re2 + (swir1 - re2) * (
        (WL_NIR - WL_RE2) / (WL_SWIR1 - WL_RE2)
    )
    fdi = nir - nir_bl_fdi

    # NDVI
    ndvi = (nir - red) / (nir + red + eps)

    # NDWI — Gao 1996
    ndwi = (green - nir) / (green + nir + eps)

    # FAI — Hu 2009
    nir_bl_fai = red + (swir1 - red) * (
        (WL_NIR - WL_RED) / (WL_SWIR1 - WL_RED)
    )
    fai = nir - nir_bl_fai

    return np.stack([fdi, ndvi, ndwi, fai], axis=0)

_d = np.random.rand(11, 256, 256).astype(np.float32)
_i = compute_spectral_indices(_d)
assert _i.shape == (4, 256, 256)
assert not np.isnan(_i).any()
print("Spectral indices OK ")
```

```python
# ============================================================
# Patch Loading Pipeline
# Purpose: Load multispectral image patches, labels, and confidence maps
# ============================================================

def load_patch(patch_id: str):
    paths = PATCH_LOOKUP[patch_id]

    with rasterio.open(paths['image']) as src:
        image = src.read().astype(np.float32)

    with rasterio.open(paths['label']) as src:
        label_raw = src.read(1).astype(np.int64)

    with rasterio.open(paths['conf']) as src:
        conf_raw = src.read(1).astype(np.int64)

    label_raw = np.clip(label_raw, 0, 15)
    label     = MERGE_LUT[label_raw]

    conf_raw  = np.clip(conf_raw, 0, 3)
    conf_map  = CONF_LUT[conf_raw]

    label[conf_map == 0.0] = -1

    return image, label, conf_map

img, lbl, conf = load_patch(train_ids[0])
print(f"Image    : {img.shape}  dtype={img.dtype}")
print(f"Label    : {lbl.shape}  unique={np.unique(lbl).tolist()}")
print(f"Conf map : {conf.shape}  unique={np.unique(conf).tolist()}")
assert (conf[lbl == -1] == 0.0).all(), "Unlabeled pixels have nonzero conf"
print("load_patch OK ")
```

```python
# ============================================================
# Normalization Statistics
# Purpose: Compute and load channel-wise normalization statistics
# ============================================================

def load_stats():
    with open(STATS_FILE) as f:
        s = json.load(f)
    means = np.array(s['means'], dtype=np.float32)
    stds  = np.array(s['stds'],  dtype=np.float32)
    assert len(means) == 15 and len(stds) == 15
    return means, stds


def normalise(full_image, means, stds):
    m = means[:, None, None]
    s = stds[:,  None, None]
    return ((full_image - m) / s).astype(np.float32)


def compute_train_stats():
    n_ch  = 15
    count = np.zeros(n_ch, dtype=np.float64)
    mean  = np.zeros(n_ch, dtype=np.float64)
    M2    = np.zeros(n_ch, dtype=np.float64)

    print(f"Computing stats over {len(train_ids)} train patches...")
    for pid in tqdm(train_ids):
        image, label, _ = load_patch(pid)
        indices    = compute_spectral_indices(image)
        full_image = np.concatenate([image, indices], axis=0)
        valid      = (label != -1)
        if valid.sum() == 0:
            continue
        for c in range(n_ch):
            vals       = full_image[c][valid].astype(np.float64)
            n          = len(vals)
            if n == 0:
                continue
            count[c]  += n
            delta      = vals - mean[c]
            mean[c]   += delta.sum() / count[c]
            delta2     = vals - mean[c]
            M2[c]     += (delta * delta2).sum()

    variance = M2 / np.maximum(count - 1, 1)
    std      = np.sqrt(variance)
    std      = np.where(std < 1e-8, 1.0, std)
    return mean, std


if STATS_FILE.exists():
    print(f"Stats file found — loading from {STATS_FILE}")
    MEANS, STDS = load_stats()
else:
    print("Stats file not found — computing...")
    MEANS, STDS = compute_train_stats()
    stats = {
        'means':         MEANS.tolist(),
        'stds':          STDS.tolist(),
        'n_channels':    15,
        'channel_names': BAND_NAMES + ['FDI','NDVI','NDWI','FAI'],
    }
    with open(STATS_FILE, 'w') as f:
        json.dump(stats, f, indent=2)
    print(f"Saved to {STATS_FILE}")

print(f"\n{'Channel':<8} {'Mean':>12} {'Std':>12}")
print("-" * 34)
for name, m, s in zip(BAND_NAMES + ['FDI','NDVI','NDWI','FAI'], MEANS, STDS):
    print(f"{name:<8} {m:>12.5f} {s:>12.5f}")
```

```python
# ============================================================
# Dataset Distribution Analysis
# Purpose: Analyze class imbalance and debris pixel distribution
# ============================================================

def analyse_distribution(patch_ids, split_name):
    class_counts  = defaultdict(int)
    debris_counts = []
    for pid in tqdm(patch_ids, desc=split_name):
        _, label, _ = load_patch(pid)
        for c in range(NUM_CLASSES):
            class_counts[c] += int((label == c).sum())
        debris_counts.append(int((label == 0).sum()))
    return class_counts, debris_counts

train_counts, train_debris = analyse_distribution(train_ids, 'train')
val_counts,   val_debris   = analyse_distribution(val_ids,   'val')
test_counts,  test_debris  = analyse_distribution(test_ids,  'test')

print(f"\n{'Class':<25} {'Train':>10} {'Val':>10} {'Test':>10}")
print("-" * 58)
for c in range(NUM_CLASSES):
    print(f"{CLASS_NAMES[c]:<25} {train_counts[c]:>10,} "
          f"{val_counts[c]:>10,} {test_counts[c]:>10,}")

total = sum(train_counts.values())
print(f"\nTotal labeled train pixels : {total:,}")
print(f"Marine Debris train pixels : {train_counts[0]:,}")
print(f"Debris % of labeled        : {100*train_counts[0]/max(total,1):.4f}%")
```

```python
# ============================================================
# Data Augmentation Pipeline
# Purpose: Define multispectral data augmentation transformations
# ============================================================

def get_train_transform():
    return A.Compose([
        A.RandomRotate90(p=0.5),
        A.HorizontalFlip(p=0.5),
        A.VerticalFlip(p=0.5),
    ], additional_targets={
        'label':    'mask',
        'conf_map': 'mask',
    })

print("Augmentation defined")
```

```python
# ============================================================
# MARIDA Dataset Pipeline
# Purpose: Define multispectral dataset preprocessing and tensor pipeline
# ============================================================

class MARIDADataset(Dataset):
    def __init__(self, patch_ids, means, stds, transform=None):
        self.patch_ids = patch_ids
        self.means     = means
        self.stds      = stds
        self.transform = transform

    def __len__(self):
        return len(self.patch_ids)

    def __getitem__(self, idx):
        pid = self.patch_ids[idx]
        image, label, conf_map = load_patch(pid)

        indices    = compute_spectral_indices(image)
        full_image = np.concatenate([image, indices], axis=0)

        if self.transform is not None:
            aug        = self.transform(
                image    = full_image.transpose(1, 2, 0),
                label    = label.astype(np.int32),
                conf_map = conf_map,
            )
            full_image = aug['image'].transpose(2, 0, 1)
            label      = aug['label'].astype(np.int64)
            conf_map   = aug['conf_map'].astype(np.float32)

        full_image = normalise(full_image, self.means, self.stds)

        return (
            torch.from_numpy(full_image),
            torch.from_numpy(np.array(label, dtype=np.int64)).long(),
            torch.from_numpy(np.array(conf_map, dtype=np.float32)),
        )

print("MARIDADataset defined")
```

```python
# ============================================================
# Weighted Sampling Strategy
# Purpose: Build weighted sampler for class imbalance handling
# ============================================================
def build_weighted_sampler(patch_ids):
    weights = []
    print(f"Building sampler for {len(patch_ids)} patches...")
    for pid in tqdm(patch_ids):
        paths = PATCH_LOOKUP[pid]
        with rasterio.open(paths['label']) as src:
            label_raw = src.read(1).astype(np.int64)
        with rasterio.open(paths['conf']) as src:
            conf_raw  = src.read(1).astype(np.int64)

        label    = MERGE_LUT[np.clip(label_raw, 0, 15)]
        conf_map = CONF_LUT[np.clip(conf_raw,  0,  3)]

        debris_mask  = (label == 0)
        debris_count = debris_mask.sum()

        if debris_count == 0:
            w = 1.0
        else:
            mean_conf = conf_map[debris_mask].mean()
            high_conf = mean_conf >= 0.75
            if debris_count <= 5:
                w = 15.0 if high_conf else 10.0
            elif debris_count <= 20:
                w = 25.0 if high_conf else 18.0
            else:
                w = 35.0   

        weights.append(w)

    return WeightedRandomSampler(
        weights=weights,
        num_samples=len(weights),
        replacement=True,
    )

print("build_weighted_sampler defined")
```

```python
# ============================================================
# Evaluation Metrics
# Purpose: Compute segmentation performance metrics for model evaluation
# ============================================================

def compute_metrics(all_preds, all_targets,
                    num_classes=NUM_CLASSES, ignore_index=-1):
    valid   = all_targets != ignore_index
    preds   = all_preds[valid]
    targets = all_targets[valid]

    if len(preds) == 0:
        return {
            'debris_f1':    0.0,
            'macro_f1':     0.0,
            'accuracy':     0.0,
            'per_class_f1': {i: 0.0 for i in range(num_classes)},
        }

    labels    = list(range(num_classes))
    debris_f1 = f1_score(targets, preds, labels=[0],
                         average='macro', zero_division=0)
    macro_f1  = f1_score(targets, preds, labels=labels,
                         average='macro', zero_division=0)
    accuracy  = accuracy_score(targets, preds)
    per_class = f1_score(targets, preds, labels=labels,
                         average=None, zero_division=0)
    return {
        'debris_f1':    float(debris_f1),
        'macro_f1':     float(macro_f1),
        'accuracy':     float(accuracy),
        'per_class_f1': {i: float(per_class[i]) for i in range(num_classes)},
    }

print("compute_metrics defined")
```

```python
# ============================================================
# Inverse Frequency Weighted Cross-Entropy Loss
# Purpose: Define weighted segmentation loss for class imbalance mitigation
# ============================================================

class InvFreqWCE(nn.Module):
    def __init__(self, class_counts: dict, ignore_index: int = -1):
        super().__init__()
        total = sum(class_counts.values())
        raw_w = torch.tensor([
            total / (2.0 * max(class_counts.get(i, 1), 1))
            for i in range(NUM_CLASSES)
        ], dtype=torch.float32)
        raw_w = raw_w / raw_w.sum() * NUM_CLASSES
        self.ce = nn.CrossEntropyLoss(
            weight=raw_w, ignore_index=ignore_index)

    def forward(self, logits, targets, conf_map=None):
        return self.ce(logits.float(), targets)

# verify
_dummy = {i: 1000 if i != 0 else 10 for i in range(NUM_CLASSES)}
_loss  = InvFreqWCE(_dummy)(
    torch.randn(2, NUM_CLASSES, 64, 64),
    torch.randint(0, NUM_CLASSES, (2, 64, 64)))
assert not torch.isnan(_loss)
print(f"InvFreqWCE OK  loss={_loss.item():.4f}")

criterion = InvFreqWCE(train_counts).to(DEVICE)
print("Criterion ready")
```

## Model Architectures

```python
# ============================================================
# Core Building Blocks
# Purpose: Define reusable convolution, residual, attention, and CBAM modules
# ============================================================

class DoubleConv(nn.Module):
    def __init__(self, in_ch, out_ch):
        super().__init__()
        self.net = nn.Sequential(
            nn.Conv2d(in_ch, out_ch, 3, padding=1, bias=False),
            nn.BatchNorm2d(out_ch), nn.ReLU(inplace=True),
            nn.Dropout2d(0.1),
            nn.Conv2d(out_ch, out_ch, 3, padding=1, bias=False),
            nn.BatchNorm2d(out_ch), nn.ReLU(inplace=True),
        )
    def forward(self, x): return self.net(x)  

class ResidualBlock(nn.Module):
    def __init__(self, channels):
        super().__init__()
        self.net = nn.Sequential(
            nn.Conv2d(channels, channels, 3, padding=1, bias=False),
            nn.BatchNorm2d(channels), nn.ReLU(inplace=True),
            nn.Conv2d(channels, channels, 3, padding=1, bias=False),
            nn.BatchNorm2d(channels),
        )
        self.relu = nn.ReLU(inplace=True)
    def forward(self, x): return self.relu(self.net(x) + x)


class AttentionGate(nn.Module):
    def __init__(self, F_g, F_l, F_int):
        super().__init__()
        self.W_g = nn.Sequential(
            nn.Conv2d(F_g, F_int, 1, bias=False), nn.BatchNorm2d(F_int))
        self.W_x = nn.Sequential(
            nn.Conv2d(F_l, F_int, 1, bias=False), nn.BatchNorm2d(F_int))
        self.psi = nn.Sequential(
            nn.Conv2d(F_int, 1, 1, bias=False),
            nn.BatchNorm2d(1), nn.Sigmoid())
        self.relu = nn.ReLU(inplace=True)
    def forward(self, g, x):
        g_up = F.interpolate(g, size=x.shape[2:],
                             mode='bilinear', align_corners=False)
        return x * self.psi(self.relu(self.W_g(g_up) + self.W_x(x)))


class CBAM(nn.Module):
    def __init__(self, channels, reduction=16):
        super().__init__()
        self.channel_att = nn.Sequential(
            nn.AdaptiveAvgPool2d(1),
            nn.Flatten(),
            nn.Linear(channels, max(channels // reduction, 1)),
            nn.ReLU(inplace=True),
            nn.Linear(max(channels // reduction, 1), channels),
            nn.Sigmoid(),
        )
        self.spatial_att = nn.Sequential(
            nn.Conv2d(2, 1, kernel_size=7, padding=3, bias=False),
            nn.Sigmoid(),
        )
    def forward(self, x):
        ca = self.channel_att(x).view(x.size(0), x.size(1), 1, 1)
        x  = x * ca
        avg_p = x.mean(dim=1, keepdim=True)
        max_p = x.max(dim=1, keepdim=True)[0]
        sa = self.spatial_att(torch.cat([avg_p, max_p], dim=1))
        return x * sa


print("Building blocks OK")
```

```python
# ============================================================
# U-Net Architecture
# Purpose: Define baseline U-Net segmentation architecture
# ============================================================
class CBAM(nn.Module):
    """Channel + Spatial attention — from SADMA paper (Mohammed 2022)."""
    def __init__(self, channels, reduction=16):
        super().__init__()
        self.channel_att = nn.Sequential(
            nn.AdaptiveAvgPool2d(1),
            nn.Flatten(),
            nn.Linear(channels, max(channels // reduction, 1)),
            nn.ReLU(inplace=True),
            nn.Linear(max(channels // reduction, 1), channels),
            nn.Sigmoid(),
        )
        self.spatial_att = nn.Sequential(
            nn.Conv2d(2, 1, kernel_size=7, padding=3, bias=False),
            nn.Sigmoid(),
        )

    def forward(self, x):
        ca = self.channel_att(x).view(x.size(0), x.size(1), 1, 1)
        x  = x * ca
        avg_p = x.mean(dim=1, keepdim=True)
        max_p = x.max(dim=1, keepdim=True)[0]
        sa = self.spatial_att(torch.cat([avg_p, max_p], dim=1))
        return x * sa


class ResAttUNet_CBAM(nn.Module):
    def __init__(self, in_channels=15, num_classes=NUM_CLASSES,
                 features=(64, 128, 256, 512)):
        super().__init__()
        self.pool = nn.MaxPool2d(2, 2)

        # Encoder — ResConv + CBAM
        self.enc_convs = nn.ModuleList()
        self.enc_cbams = nn.ModuleList()
        ch = in_channels
        for f in features:
            self.enc_convs.append(self._res_block(ch, f))
            self.enc_cbams.append(CBAM(f))
            ch = f

        # Bottleneck
        self.bottleneck = self._res_block(features[-1], features[-1] * 2)

        # Decoder — transpose conv + attention gate + ResConv
        self.up_convs  = nn.ModuleList()
        self.att_gates = nn.ModuleList()
        self.dec_convs = nn.ModuleList()
        for f in reversed(features):
            self.up_convs.append(
                nn.ConvTranspose2d(f * 2, f, kernel_size=2, stride=2))
            self.att_gates.append(
                AttentionGate(F_g=f, F_l=f, F_int=f // 2))
            self.dec_convs.append(self._res_block(f * 2, f))

        self.final_conv = nn.Conv2d(features[0], num_classes, 1)

    @staticmethod
    def _res_block(in_ch, out_ch):
        """Conv-BN-ReLU x2 with residual projection."""
        conv = nn.Sequential(
            nn.Conv2d(in_ch, out_ch, 3, padding=1, bias=False),
            nn.BatchNorm2d(out_ch), nn.ReLU(inplace=True),
            nn.Conv2d(out_ch, out_ch, 3, padding=1, bias=False),
            nn.BatchNorm2d(out_ch),
        )
        proj = (nn.Conv2d(in_ch, out_ch, 1, bias=False)
                if in_ch != out_ch else nn.Identity())
        relu = nn.ReLU(inplace=True)

        class Block(nn.Module):
            def forward(self, x):
                return relu(conv(x) + proj(x))
        b = Block(); b.conv = conv; b.proj = proj; b.relu = relu
        return b

    def forward(self, x):
        skips = []
        for conv, cbam in zip(self.enc_convs, self.enc_cbams):
            x = cbam(conv(x))  
            skips.append(x)
            x = self.pool(x)

        x = self.bottleneck(x)
        skips = skips[::-1]

        for up, att, dec, skip in zip(
                self.up_convs, self.att_gates, self.dec_convs, skips):
            x = up(x)
            if x.shape != skip.shape:
                x = F.interpolate(x, size=skip.shape[2:],
                                  mode='bilinear', align_corners=False)
            skip = att(g=x, x=skip)  
            x = dec(torch.cat([skip, x], dim=1))

        return self.final_conv(x)



_y = ResAttUNet_CBAM()(torch.randn(2, 15, 256, 256))
assert _y.shape == (2, NUM_CLASSES, 256, 256)
print(f"ResAttUNet_CBAM OK   output: {_y.shape}")
total_params = sum(p.numel() for p in ResAttUNet_CBAM().parameters())
print(f"Parameters: {total_params/1e6:.1f}M")
```

```python
# ============================================================
# U-Net Architecture
# Purpose: Define baseline U-Net segmentation architecture
# ============================================================

class UNet(nn.Module):
    def __init__(self, in_channels=15, num_classes=NUM_CLASSES,
                 features=(64, 128, 256, 512)):
        super().__init__()
        self.downs    = nn.ModuleList()
        self.ups      = nn.ModuleList()
        self.up_convs = nn.ModuleList()
        self.pool     = nn.MaxPool2d(2, 2)
        ch = in_channels
        for f in features:
            self.downs.append(DoubleConv(ch, f)); ch = f
        self.bottleneck = DoubleConv(features[-1], features[-1] * 2)
        for f in reversed(features):
            self.up_convs.append(
                nn.ConvTranspose2d(f * 2, f, kernel_size=2, stride=2))
            self.ups.append(DoubleConv(f * 2, f))
        self.final_conv = nn.Conv2d(features[0], num_classes, 1)

    def forward(self, x):
        skips = []
        for down in self.downs:
            x = down(x); skips.append(x); x = self.pool(x)
        x = self.bottleneck(x); skips = skips[::-1]
        for i, (uc, up) in enumerate(zip(self.up_convs, self.ups)):
            x = uc(x); skip = skips[i]
            if x.shape != skip.shape:
                x = F.interpolate(x, size=skip.shape[2:],
                                  mode='bilinear', align_corners=False)
            x = up(torch.cat([skip, x], dim=1))
        return self.final_conv(x)

_y = UNet()(torch.randn(2, 15, 256, 256))
assert _y.shape == (2, NUM_CLASSES, 256, 256)
print(f"UNet OK output: {_y.shape}")
```

```python
# ============================================================
# Attention U-Net Architecture
# Purpose: Define Attention U-Net segmentation architecture
# ============================================================

class AttentionUNet(nn.Module):
    def __init__(self, in_channels=15, num_classes=NUM_CLASSES,
                 features=(64, 128, 256, 512)):
        super().__init__()
        self.downs    = nn.ModuleList()
        self.ups      = nn.ModuleList()
        self.up_convs = nn.ModuleList()
        self.attns    = nn.ModuleList()
        self.pool     = nn.MaxPool2d(2, 2)
        ch = in_channels
        for f in features:
            self.downs.append(DoubleConv(ch, f)); ch = f
        self.bottleneck = DoubleConv(features[-1], features[-1] * 2)
        for f in reversed(features):
            self.up_convs.append(
                nn.ConvTranspose2d(f * 2, f, kernel_size=2, stride=2))
            self.attns.append(AttentionGate(F_g=f, F_l=f, F_int=f // 2))
            self.ups.append(DoubleConv(f * 2, f))
        self.final_conv = nn.Conv2d(features[0], num_classes, 1)

    def forward(self, x):
        skips = []
        for down in self.downs:
            x = down(x); skips.append(x); x = self.pool(x)
        x = self.bottleneck(x); skips = skips[::-1]
        for i, (uc, attn, up) in enumerate(
                zip(self.up_convs, self.attns, self.ups)):
            x = uc(x); skip = skips[i]
            if x.shape != skip.shape:
                x = F.interpolate(x, size=skip.shape[2:],
                                  mode='bilinear', align_corners=False)
            skip = attn(g=x, x=skip)
            x = up(torch.cat([skip, x], dim=1))
        return self.final_conv(x)

_y = AttentionUNet()(torch.randn(2, 15, 256, 256))
assert _y.shape == (2, NUM_CLASSES, 256, 256)
print(f"AttentionUNet OK output: {_y.shape}")
```

```python
# ============================================================
# Training and Evaluation Loops
# Purpose: Define batch-wise model training and evaluation procedures
# ============================================================

def train_one_epoch(model, loader, optimizer, criterion,
                    device, scaler=None, clip_grad=0.5,
                    stage1_model=None):
    model.train()
    if stage1_model is not None:
        stage1_model.eval()

    total_loss = 0.0
    n_batches  = 0
    n_skipped  = 0
    all_preds, all_targets = [], []

    for images, labels, conf_maps in tqdm(loader, desc='train', leave=False):
        images    = images.to(device)
        labels    = labels.to(device)
        conf_maps = conf_maps.to(device)

        if (labels != -1).sum() == 0:
            n_skipped += 1
            continue

        if stage1_model is not None:
            with torch.no_grad():
                s1_probs    = F.softmax(stage1_model(images), dim=1)
                debris_prob = s1_probs[:, 0:1, :, :]
            images = torch.cat([images, debris_prob], dim=1)

        if torch.isnan(images).any():
            n_skipped += 1
            continue

        optimizer.zero_grad()
        logits = model(images)

        if torch.isnan(logits).any():
            optimizer.zero_grad()
            n_skipped += 1
            continue

        loss = criterion(logits, labels, conf_maps)

        if torch.isnan(loss) or torch.isinf(loss):
            optimizer.zero_grad()
            n_skipped += 1
            continue

        loss.backward()

        has_nan_grad = any(
            p.grad is not None and torch.isnan(p.grad).any()
            for p in model.parameters()
        )
        if has_nan_grad:
            optimizer.zero_grad()
            n_skipped += 1
            continue

        torch.nn.utils.clip_grad_norm_(model.parameters(), clip_grad)
        optimizer.step()

        total_loss += loss.item()
        n_batches  += 1

        with torch.no_grad():
            preds = torch.argmax(logits, dim=1)
            all_preds.append(preds.cpu().numpy().flatten())
            all_targets.append(labels.cpu().numpy().flatten())

    if n_skipped > 0:
        print(f"  Skipped {n_skipped} batches this epoch")

    avg_loss = total_loss / max(n_batches, 1)
    metrics  = compute_metrics(
        np.concatenate(all_preds)   if all_preds   else np.array([0]),
        np.concatenate(all_targets) if all_targets else np.array([0]),
    )
    return avg_loss, metrics


@torch.no_grad()
def evaluate(model, loader, criterion, device, stage1_model=None):
    model.eval()
    total_loss = 0.0
    n_batches  = 0
    all_preds, all_targets = [], []

    for images, labels, conf_maps in tqdm(loader, desc='eval', leave=False):
        images    = images.to(device)
        labels    = labels.to(device)
        conf_maps = conf_maps.to(device)

        if (labels != -1).sum() == 0:
            continue

        if stage1_model is not None:
            s1_probs    = F.softmax(stage1_model(images), dim=1)
            debris_prob = s1_probs[:, 0:1, :, :]
            images      = torch.cat([images, debris_prob], dim=1)

        logits = model(images)
        loss   = criterion(logits, labels, conf_maps)

        if not torch.isnan(loss) and not torch.isinf(loss):
            total_loss += loss.item()
            n_batches  += 1

        preds = torch.argmax(logits, dim=1)
        all_preds.append(preds.cpu().numpy().flatten())
        all_targets.append(labels.cpu().numpy().flatten())

    avg_loss = total_loss / max(n_batches, 1)
    metrics  = compute_metrics(
        np.concatenate(all_preds)   if all_preds   else np.array([0]),
        np.concatenate(all_targets) if all_targets else np.array([0]),
    )
    return avg_loss, metrics

print("Training loop defined")
```

```python
# ============================================================
# Training Pipeline
# Purpose: Define training, validation, checkpointing, and optimization workflow
# ============================================================

import math
def run_training(model, train_loader, val_loader, model_name,
                 epochs=50, lr=5e-4, weight_decay=1e-5,
                 clip_grad=0.5, stage1_model=None, device=DEVICE):

    model     = model.to(device)
    criterion = InvFreqWCE(train_counts).to(device)
    optimizer = torch.optim.Adam(
        model.parameters(), lr=lr, weight_decay=weight_decay)
    def warmup_cosine(epoch):
        if epoch < 5:
            return (epoch + 1) / 5        # linear warmup for 5 epochs
        progress = (epoch - 5) / (epochs - 5)
        return 0.5 * (1 + math.cos(math.pi * progress))

    scheduler = torch.optim.lr_scheduler.LambdaLR(optimizer, warmup_cosine)
    scaler    = GradScaler() if device.type == 'cuda' else None

    ckpt_path    = CKPT_DIR / f'{model_name}_best.pth'
    best_f1      = -1.0
    best_metrics = {}
    history = {
        'train_loss': [], 'val_loss': [],
        'train_debris_f1': [], 'val_debris_f1': [],
        'train_macro_f1':  [], 'val_macro_f1':  [],
    }

    print(f"\n{'='*60}")
    print(f"Training: {model_name}  epochs={epochs}  device={device}")
    print(f"{'='*60}")

    for epoch in range(1, epochs + 1):
        tr_loss, tr_m = train_one_epoch(
            model, train_loader, optimizer, criterion,
            device, scaler, clip_grad, stage1_model)
        va_loss, va_m = evaluate(
            model, val_loader, criterion, device, stage1_model)
        scheduler.step()

        history['train_loss'].append(tr_loss)
        history['val_loss'].append(va_loss)
        history['train_debris_f1'].append(tr_m['debris_f1'])
        history['val_debris_f1'].append(va_m['debris_f1'])
        history['train_macro_f1'].append(tr_m['macro_f1'])
        history['val_macro_f1'].append(va_m['macro_f1'])

        flag = ''
        if va_m['debris_f1'] > best_f1:
            best_f1      = va_m['debris_f1']
            best_metrics = va_m
            torch.save({
                'epoch':     epoch,
                'model':     model.state_dict(),
                'optimizer': optimizer.state_dict(),
                'scheduler': scheduler.state_dict(),
                'metrics':   va_m,
            }, ckpt_path)
            flag = ' ← best'

        print(f"Ep {epoch:3d}/{epochs} | "
              f"Loss {tr_loss:.4f}/{va_loss:.4f} | "
              f"DebrisF1 {tr_m['debris_f1']:.3f}/{va_m['debris_f1']:.3f} | "
              f"MacroF1 {va_m['macro_f1']:.3f}{flag}")

    print(f"\nBest val Debris F1 : {best_f1:.4f}")
    print(f"Checkpoint         : {ckpt_path}")
    return history, best_metrics

print("run_training defined")
```

```python
# =============================================================
# SpectralAttUNet
# 1. ResAttUNet (Mohammed 2022)  => CBAM + residual blocks
# 2. MariNeXt (Kikaki 2024)     => convolutional attention
# 3. Kikaki 2022                 => FDI+NDVI+NDWI most important
# Novel contribution: dedicated spectral index branch
# fused into decoder at each stage
#=============================================================
import gc, math, os
gc.collect(); torch.cuda.empty_cache()

class SpectralBranch(nn.Module):
    """
    Dedicated branch for spectral indices (FDI, NDVI, NDWI, FAI)
    Paper: Kikaki 2022 found these 4 indices most important
    Processes indices separately then fuses with main decoder
    """
    def __init__(self, n_indices=4, out_ch=64):
        super().__init__()
        self.net = nn.Sequential(
            nn.Conv2d(n_indices, 32, 3, padding=1, bias=False),
            nn.BatchNorm2d(32), nn.ReLU(inplace=True),
            nn.Conv2d(32, out_ch, 3, padding=1, bias=False),
            nn.BatchNorm2d(out_ch), nn.ReLU(inplace=True),
        )
    def forward(self, x): return self.net(x)


class ResConvBlock(nn.Module):
    """
    Residual conv block — from ResAttUNet paper
    Prevents gradient vanishing on rare debris pixels
    """
    def __init__(self, in_ch, out_ch):
        super().__init__()
        self.conv = nn.Sequential(
            nn.Conv2d(in_ch, out_ch, 3, padding=1, bias=False),
            nn.BatchNorm2d(out_ch), nn.ReLU(inplace=True),
            nn.Conv2d(out_ch, out_ch, 3, padding=1, bias=False),
            nn.BatchNorm2d(out_ch),
        )
        self.skip = (nn.Conv2d(in_ch, out_ch, 1, bias=False)
                     if in_ch != out_ch else nn.Identity())
        self.relu = nn.ReLU(inplace=True)
        self.drop = nn.Dropout2d(0.05)

    def forward(self, x):
        return self.relu(self.drop(self.conv(x)) + self.skip(x))


class SpectralAttUNet(nn.Module):
    """
    SpectralAttUNet — novel proposed model
    
    Architecture sources:
    - Encoder: ResConvBlock + CBAM (Mohammed 2022)
    - Bottleneck: 3 residual blocks (Mohammed 2022)
    - Decoder: CBAM on skip connections (Mohammed 2022)
    - Spectral branch: FDI/NDVI/NDWI/FAI fusion (Kikaki 2022)
    - Attention gates: original Attention UNet (Oktay 2018)
    
    Novel: spectral index branch fused at decoder stage 1
    No paper on MARIDA has done this combination
    """
    def __init__(self, in_channels=15, num_classes=NUM_CLASSES,
                 features=(64, 128, 256, 512)):
        super().__init__()
        self.pool = nn.MaxPool2d(2, 2)

        # Encoder — ResConvBlock + CBAM (like ResAttUNet)
        self.enc = nn.ModuleList()
        self.enc_cbam = nn.ModuleList()
        ch = in_channels
        for f in features:
            self.enc.append(ResConvBlock(ch, f))
            self.enc_cbam.append(CBAM(f))
            ch = f

        # Bottleneck — 3 residual blocks (Mohammed 2022 key detail)
        self.bottleneck = nn.Sequential(
            ResConvBlock(features[-1], features[-1] * 2),
            ResConvBlock(features[-1] * 2, features[-1] * 2),
            ResConvBlock(features[-1] * 2, features[-1] * 2),
        )

        # spectral index branch — processes FDI,NDVI,NDWI,FAI
        # channels 11-14 in your 15-channel input
        self.spectral_branch = SpectralBranch(
            n_indices=4, out_ch=features[0])

        # Decoder — transpose conv + attention + CBAM + ResConv
        self.up_convs  = nn.ModuleList()
        self.att_gates = nn.ModuleList()
        self.dec_cbam  = nn.ModuleList()
        self.dec       = nn.ModuleList()

        rev_features = list(reversed(features))
        for i, f in enumerate(rev_features):
            self.up_convs.append(
                nn.ConvTranspose2d(f * 2, f, 2, stride=2))
            self.att_gates.append(
                AttentionGate(F_g=f, F_l=f, F_int=f // 2))
            self.dec_cbam.append(CBAM(f))
            # last decoder level: +spectral branch channels
            extra = features[0] if i == len(rev_features)-1 else 0
            self.dec.append(ResConvBlock(f * 2 + extra, f))

        self.final_conv = nn.Conv2d(features[0], num_classes, 1)

    def forward(self, x):
        spectral = self.spectral_branch(x[:, 11:15, :, :])

        # Encoder
        skips = []
        for conv, cbam in zip(self.enc, self.enc_cbam):
            x = cbam(conv(x))
            skips.append(x)
            x = self.pool(x)

        # Bottleneck
        x = self.bottleneck(x)
        skips = skips[::-1]

        # Decoder
        for i, (up, att, dcbam, dec) in enumerate(
                zip(self.up_convs, self.att_gates,
                    self.dec_cbam, self.dec)):
            x   = up(x)
            skip = skips[i]
            if x.shape != skip.shape:
                x = F.interpolate(x, size=skip.shape[2:],
                                  mode='bilinear',
                                  align_corners=False)
            skip = att(g=x, x=skip)
            skip = dcbam(skip)

            if i == len(self.up_convs) - 1:
                if spectral.shape != x.shape:
                    spectral = F.interpolate(
                        spectral, size=x.shape[2:],
                        mode='bilinear', align_corners=False)
                x = dec(torch.cat([skip, x, spectral], dim=1))
            else:
                x = dec(torch.cat([skip, x], dim=1))

        return self.final_conv(x)


gc.collect(); torch.cuda.empty_cache()
_model = SpectralAttUNet(
    in_channels=15, num_classes=NUM_CLASSES).to(DEVICE)
_x = torch.randn(2, 15, 256, 256).to(DEVICE)
with torch.no_grad():
    _y = _model(_x)
assert _y.shape == (2, NUM_CLASSES, 256, 256)
assert not torch.isnan(_y).any()
n = sum(p.numel() for p in _model.parameters()) / 1e6
print(f"SpectralAttUNet OK output: {_y.shape}")
print(f"Parameters: {n:.1f}M")
print(f"Output range: {_y.min():.2f} to {_y.max():.2f}")
del _x, _y, _model
torch.cuda.empty_cache()
```

```python
# ============================================================
# DeepSpectralUNet Architecture
# Purpose: Define the proposed DeepSpectralUNet segmentation architecture
# ============================================================

class MultiScaleSpectralBranch(nn.Module):
    def __init__(self, n_indices=4, features=(64, 128, 256, 512)):
        super().__init__()
        self.scales = nn.ModuleList()
        self.pool   = nn.MaxPool2d(2, 2)
        in_ch = n_indices
        for f in features:
            self.scales.append(nn.Sequential(
                nn.Conv2d(in_ch, f, 3, padding=1, bias=False),
                nn.BatchNorm2d(f), nn.ReLU(inplace=True),
            ))
            in_ch = f

    def forward(self, x):
        outputs = []
        for scale in self.scales:
            x = scale(x)
            outputs.append(x)
            x = self.pool(x)
        return outputs


class DeepSpectralUNet(nn.Module):
    """
    DeepSpectralUNet — improved proposed model.
    Novel contribution: multi-scale spectral index branch
    fused at every decoder stage (vs single-stage in SpectralAttUNet).
    Sources: Mohammed 2022, Kikaki 2022, Oktay 2018
    """
    def __init__(self, in_channels=15, num_classes=NUM_CLASSES,
                 features=(64, 128, 256, 512)):
        super().__init__()
        self.pool = nn.MaxPool2d(2, 2)

        self.enc      = nn.ModuleList()
        self.enc_cbam = nn.ModuleList()
        ch = in_channels
        for f in features:
            self.enc.append(ResConvBlock(ch, f))
            self.enc_cbam.append(CBAM(f))
            ch = f

        self.bottleneck = nn.Sequential(
            ResConvBlock(features[-1], features[-1] * 2),
            ResConvBlock(features[-1] * 2, features[-1] * 2),
            ResConvBlock(features[-1] * 2, features[-1] * 2),
        )

        self.spectral_branch = MultiScaleSpectralBranch(
            n_indices=4, features=features)

        self.up_convs  = nn.ModuleList()
        self.att_gates = nn.ModuleList()
        self.dec_cbam  = nn.ModuleList()
        self.dec       = nn.ModuleList()
        for f in reversed(features):
            self.up_convs.append(
                nn.ConvTranspose2d(f * 2, f, 2, stride=2))
            self.att_gates.append(
                AttentionGate(F_g=f, F_l=f, F_int=f // 2))
            self.dec_cbam.append(CBAM(f))
            self.dec.append(ResConvBlock(f * 2 + f, f))

        self.final_conv = nn.Conv2d(features[0], num_classes, 1)

    def forward(self, x):
        spectral_maps = self.spectral_branch(x[:, 11:15, :, :])
        spectral_maps = spectral_maps[::-1]

        skips = []
        for conv, cbam in zip(self.enc, self.enc_cbam):
            x = cbam(conv(x))
            skips.append(x)
            x = self.pool(x)

        x = self.bottleneck(x)
        skips = skips[::-1]

        for i, (up, att, dcbam, dec) in enumerate(
                zip(self.up_convs, self.att_gates,
                    self.dec_cbam, self.dec)):
            x    = up(x)
            skip = skips[i]
            if x.shape != skip.shape:
                x = F.interpolate(x, size=skip.shape[2:],
                                  mode='bilinear', align_corners=False)
            skip = att(g=x, x=skip)
            skip = dcbam(skip)
            spec = spectral_maps[i]
            if spec.shape[2:] != x.shape[2:]:
                spec = F.interpolate(spec, size=x.shape[2:],
                                     mode='bilinear', align_corners=False)
            x = dec(torch.cat([skip, x, spec], dim=1))

        return self.final_conv(x)


# sanity check
gc.collect(); torch.cuda.empty_cache()
_model = DeepSpectralUNet(in_channels=15, num_classes=NUM_CLASSES).to(DEVICE)
_x     = torch.randn(2, 15, 256, 256).to(DEVICE)
with torch.no_grad():
    _y = _model(_x)
assert _y.shape == (2, NUM_CLASSES, 256, 256)
assert not torch.isnan(_y).any()
n = sum(p.numel() for p in _model.parameters()) / 1e6
print(f"DeepSpectralUNet OK output: {_y.shape}")
print(f"Parameters: {n:.1f}M")
del _x, _y, _model
torch.cuda.empty_cache()
```

```python
# ============================================================
# Figure 1: Prediction Visualization
# Purpose: Visualize segmentation predictions from all trained models
# ============================================================
import gc
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import matplotlib.colors as mcolors

gc.collect(); torch.cuda.empty_cache()

COLORS = ['red','limegreen','blue','yellow','white',
          'navy','cyan','magenta','gray','orange']

def make_rgb(image):
    rgb = np.stack([image[3], image[2], image[1]], axis=2)
    return (rgb - rgb.min()) / (rgb.max() - rgb.min() + 1e-8)

def make_vis(mask):
    vis = np.zeros((*mask.shape, 3))
    for cls, col in enumerate(COLORS):
        vis[mask == cls] = mcolors.to_rgb(col)
    return vis

# find patch with most debris
best_pid, best_count = None, 0
for pid in test_ids:
    _, label, _ = load_patch(pid)
    if (label == 0).sum() > best_count:
        best_count = (label == 0).sum()
        best_pid = pid

print(f"Best patch: {best_pid} with {best_count} debris pixels")

# prepare input
image, label, _ = load_patch(best_pid)
indices  = compute_spectral_indices(image)
full_img = normalise(
    np.concatenate([image, indices], axis=0), MEANS, STDS)
inp = torch.from_numpy(full_img).unsqueeze(0).to(DEVICE)

# get predictions from all models
kw = dict(in_channels=15, num_classes=NUM_CLASSES)
MODELS = [
    ('U-Net',            'unet_fixed_best.pth',       UNet),
    ('Attention U-Net',  'attunet_fixed_best.pth',     AttentionUNet),
    ('DeepLabV3+',       'deeplabv3plus_best.pth',     DeepLabV3PlusBaseline),
    ('ResAttUNet-CBAM',  'resattunet_cbam_best.pth',   ResAttUNet_CBAM),
    ('SpectralAttUNet',  'spectral_attunet_best.pth',  SpectralAttUNet),
    ('DeepSpectralUNet', 'deep_spectral_unet_best.pth',DeepSpectralUNet),
]

predictions = {}
for name, fname, ModelClass in MODELS:
    ckpt_path = CKPT_DIR / fname
    if not ckpt_path.exists():
        print(f"Missing: {fname}")
        continue
    m = ModelClass(**kw).to(DEVICE)
    ckpt = torch.load(ckpt_path, map_location=DEVICE,
                      weights_only=False)
    m.load_state_dict(ckpt.get('model', ckpt))
    m.eval()
    with torch.no_grad():
        pred = torch.argmax(m(inp), dim=1)[0].cpu().numpy()
    predictions[name] = pred
    del m; gc.collect(); torch.cuda.empty_cache()
    print(f"Done: {name}")

# plot
n = len(predictions)
fig, axes = plt.subplots(2, n+1, figsize=(4*(n+1), 8))

# top row — RGB and ground truth
axes[0][0].imshow(make_rgb(image))
axes[0][0].set_title('RGB Input', fontweight='bold', fontsize=9)
axes[0][0].axis('off')

axes[1][0].imshow(make_vis(label))
axes[1][0].set_title('Ground Truth', fontweight='bold', fontsize=9)
axes[1][0].axis('off')

# predictions
for i, (name, pred) in enumerate(predictions.items()):
    axes[0][i+1].imshow(make_vis(pred))
    axes[0][i+1].set_title(name.replace(' ','\n'),
                             fontweight='bold', fontsize=8)
    axes[0][i+1].axis('off')
    axes[1][i+1].axis('off')

# legend
patches = [mpatches.Patch(color=COLORS[i],
           label=CLASS_NAMES[i]) for i in range(NUM_CLASSES)]
fig.legend(handles=patches, loc='lower center',
           ncol=5, fontsize=8, bbox_to_anchor=(0.5, -0.02))
plt.suptitle(f'Prediction Maps — {best_pid}\n'
             f'(debris pixels: {best_count})',
             fontsize=11, fontweight='bold')
plt.tight_layout()
plt.savefig(RESULTS_DIR/'fig1_prediction_maps.png',
            dpi=150, bbox_inches='tight')
plt.show()
print('Saved fig1_prediction_maps.png')
```

```python
# ============================================================
# Figure 2: Confusion Matrix
# Purpose: Generate normalized confusion matrix for DeepSpectralUNet predictions
# ============================================================
import gc
from sklearn.metrics import confusion_matrix
import seaborn as sns

gc.collect(); torch.cuda.empty_cache()

ds_model = DeepSpectralUNet(
    in_channels=15, num_classes=NUM_CLASSES).to(DEVICE)
ckpt = torch.load(CKPT_DIR/'deep_spectral_unet_best.pth',
                  map_location=DEVICE, weights_only=False)
ds_model.load_state_dict(ckpt.get('model', ckpt))
ds_model.eval()

vp, vt = [], []
with torch.no_grad():
    for images, labels, _ in tqdm(
            test_loader, desc='CM', leave=False):
        preds = torch.argmax(
            ds_model(images.to(DEVICE)), dim=1)
        mask = labels != -1
        vp.append(preds[mask].cpu().numpy())
        vt.append(labels[mask].numpy())

cm = confusion_matrix(
    np.concatenate(vt), np.concatenate(vp),
    labels=list(range(NUM_CLASSES)))
cm_norm = cm.astype(float) / (
    cm.sum(axis=1, keepdims=True) + 1e-8)

plt.figure(figsize=(12, 10))
sns.heatmap(cm_norm, annot=True, fmt='.2f',
            cmap='Blues',
            xticklabels=CLASS_NAMES,
            yticklabels=CLASS_NAMES)
plt.title('DeepSpectralUNet — Normalised Confusion Matrix',
          fontweight='bold', fontsize=12)
plt.ylabel('True Label')
plt.xlabel('Predicted Label')
plt.tight_layout()
plt.savefig(RESULTS_DIR/'fig2_confusion_matrix.png',
            dpi=150, bbox_inches='tight')
plt.show()
print('Saved fig2_confusion_matrix.png')
del ds_model; gc.collect(); torch.cuda.empty_cache()
```

```python
# ============================================================
# Figure 3: Per-Class IoU Comparison
# Purpose: Compute and visualize per-class IoU scores for all trained models
# ============================================================
import gc
from sklearn.metrics import f1_score as skf1

gc.collect(); torch.cuda.empty_cache()

def get_per_class_iou(ckpt_path, ModelClass, **kwargs):
    m = ModelClass(**kwargs).to(DEVICE)
    ckpt = torch.load(ckpt_path,
                      map_location=DEVICE,
                      weights_only=False)
    m.load_state_dict(ckpt.get('model', ckpt))
    m.eval()
    vp, vt = [], []
    with torch.no_grad():
        for images, labels, _ in test_loader:
            preds = torch.argmax(
                m(images.to(DEVICE)), dim=1)
            mask  = labels != -1
            vp.append(preds[mask].cpu().numpy())
            vt.append(labels[mask].numpy())
    del m; gc.collect(); torch.cuda.empty_cache()
    f1s = skf1(np.concatenate(vt),
                np.concatenate(vp),
                labels=list(range(NUM_CLASSES)),
                average=None, zero_division=0)
    return np.array([f1_to_iou(f) for f in f1s])

kw = dict(in_channels=15, num_classes=NUM_CLASSES)
MODELS_VIZ = [
    ('U-Net',            CKPT_DIR/'unet_fixed_best.pth',        UNet),
    ('ResAttUNet-CBAM',  CKPT_DIR/'resattunet_cbam_best.pth',   ResAttUNet_CBAM),
    ('SpectralAttUNet',  CKPT_DIR/'spectral_attunet_best.pth',  SpectralAttUNet),
    ('DeepSpectralUNet', CKPT_DIR/'deep_spectral_unet_best.pth',DeepSpectralUNet),
]

per_class_iou = {}
for name, ckpt_path, ModelClass in MODELS_VIZ:
    if not ckpt_path.exists():
        print(f"Missing: {name}")
        continue
    print(f"Computing per class IoU: {name}...")
    per_class_iou[name] = get_per_class_iou(
        ckpt_path, ModelClass, **kw)

# plot
n  = len(per_class_iou)
x  = np.arange(NUM_CLASSES)
w  = 0.2
colors = ['steelblue','green','purple','brown']

fig, ax = plt.subplots(figsize=(16, 7))
for i, (name, iou) in enumerate(per_class_iou.items()):
    offset = (i - n/2 + 0.5) * w
    ax.bar(x + offset, iou, w,
           label=name, color=colors[i], alpha=0.85)

ax.set_xticks(x)
ax.set_xticklabels(CLASS_NAMES,
                    rotation=30, ha='right', fontsize=9)
ax.set_ylabel('IoU Score', fontsize=11)
ax.set_title('Per-Class IoU Comparison — Test Set',
             fontsize=13, fontweight='bold')
ax.legend(fontsize=9)
ax.grid(axis='y', alpha=0.3)
ax.set_ylim(0, 1.0)
ax.axhline(y=0, color='black', linewidth=0.5)
plt.tight_layout()
plt.savefig(RESULTS_DIR/'fig3_per_class_iou.png',
            dpi=150, bbox_inches='tight')
plt.show()
print('Saved fig3_per_class_iou.png')
```

```python
# ============================================================
# Figure 4: Attention Gate Visualization
#
# Purpose:
# This section visualizes attention gate activations from DeepSpectralUNet during inference.
# It extracts attention maps from the model's attention gates and overlays them on the RGB input image.
# ============================================================

import gc
gc.collect(); torch.cuda.empty_cache()

attention_maps = {}
hooks = []

def make_hook(name):
    def hook(module, input, output):
        attention_maps[name] = output.detach().cpu()
    return hook

ds_model = DeepSpectralUNet(
    in_channels=15, num_classes=NUM_CLASSES).to(DEVICE)
ckpt = torch.load(CKPT_DIR/'deep_spectral_unet_best.pth',
                  map_location=DEVICE, weights_only=False)
ds_model.load_state_dict(ckpt.get('model', ckpt))
ds_model.eval()

# register hooks on attention gates psi layer
for i, att_gate in enumerate(ds_model.att_gates):
    h = att_gate.psi.register_forward_hook(
        make_hook(f'Attention Gate {i+1}'))
    hooks.append(h)

with torch.no_grad():
    _ = ds_model(inp)

for h in hooks: h.remove()

# plot
n_gates = len(attention_maps)
fig, axes = plt.subplots(
    1, n_gates + 2, figsize=(4*(n_gates+2), 4))

axes[0].imshow(make_rgb(image))
axes[0].set_title('RGB Input', fontweight='bold', fontsize=9)
axes[0].axis('off')

axes[1].imshow(make_vis(label))
axes[1].set_title('Ground Truth', fontweight='bold', fontsize=9)
axes[1].axis('off')

for i, (name, att_map) in enumerate(
        attention_maps.items()):
    att = att_map[0, 0].numpy()
    att = (att - att.min()) / (att.max() - att.min() + 1e-8)
    att_resized = F.interpolate(
        torch.from_numpy(att).unsqueeze(0).unsqueeze(0).float(),
        size=(256, 256), mode='bilinear',
        align_corners=False)[0, 0].numpy()
    axes[i+2].imshow(make_rgb(image))
    axes[i+2].imshow(att_resized, alpha=0.6, cmap='jet')
    axes[i+2].set_title(name, fontweight='bold', fontsize=8)
    axes[i+2].axis('off')

plt.suptitle('DeepSpectralUNet — Attention Gate Visualisations\n'
             'Red = high attention, Blue = low attention',
             fontsize=11, fontweight='bold')
plt.tight_layout()
plt.savefig(RESULTS_DIR/'fig4_attention_maps.png',
            dpi=150, bbox_inches='tight')
plt.show()
print('Saved fig4_attention_maps.png')
del ds_model; gc.collect(); torch.cuda.empty_cache()
```

```python
# ============================================================
# Figure 5: Spectral Branch Feature Visualization
#
# Purpose: This section visualizes intermediate feature activations learned by the spectral branch of DeepSpectralUNet.
# ============================================================
import gc
gc.collect(); torch.cuda.empty_cache()

spectral_features = {}

def spectral_hook(name):
    def hook(module, input, output):
        spectral_features[name] = output.detach().cpu()
    return hook

ds_model = DeepSpectralUNet(
    in_channels=15, num_classes=NUM_CLASSES).to(DEVICE)
ckpt = torch.load(CKPT_DIR/'deep_spectral_unet_best.pth',
                  map_location=DEVICE, weights_only=False)
ds_model.load_state_dict(ckpt.get('model', ckpt))
ds_model.eval()

# hook on each spectral scale
scale_names = ['Scale 1 — Full (256×256)',
               'Scale 2 — Half (128×128)',
               'Scale 3 — Quarter (64×64)',
               'Scale 4 — Eighth (32×32)']

for i, scale in enumerate(ds_model.spectral_branch.scales):
    scale.register_forward_hook(
        spectral_hook(scale_names[i]))

with torch.no_grad():
    _ = ds_model(inp)

# get spectral indices for display
indices_raw = inp[0, 11:15].cpu().numpy()
index_names = ['FDI', 'NDVI', 'NDWI', 'FAI']

# plot
fig, axes = plt.subplots(
    2, max(len(spectral_features), 4)+1,
    figsize=(4*(max(len(spectral_features),4)+1), 8))

# top row — raw spectral indices
for i in range(4):
    idx = indices_raw[i]
    idx_norm = (idx - idx.min()) / (idx.max() - idx.min() + 1e-8)
    axes[0][i].imshow(idx_norm, cmap='RdYlGn')
    axes[0][i].set_title(f'Input: {index_names[i]}',
                          fontweight='bold', fontsize=8)
    axes[0][i].axis('off')

# hide extra top axes
for i in range(4, axes[0].shape[0]):
    axes[0][i].axis('off')

# bottom row — learned features at each scale
for i, (name, feat) in enumerate(
        spectral_features.items()):
    avg = feat[0].mean(0).numpy()
    avg = (avg - avg.min()) / (avg.max() - avg.min() + 1e-8)
    avg_resized = F.interpolate(
        torch.from_numpy(avg).unsqueeze(0).unsqueeze(0).float(),
        size=(256, 256), mode='bilinear',
        align_corners=False)[0, 0].numpy()
    axes[1][i].imshow(avg_resized, cmap='viridis')
    axes[1][i].set_title(name, fontweight='bold', fontsize=7)
    axes[1][i].axis('off')

# hide extra bottom axes
for i in range(len(spectral_features), axes[1].shape[0]):
    axes[1][i].axis('off')

plt.suptitle('MultiScaleSpectralBranch — Input Indices vs Learned Features\n'
             'Top: Raw spectral indices | Bottom: Learned representations',
             fontsize=11, fontweight='bold')
plt.tight_layout()
plt.savefig(RESULTS_DIR/'fig5_spectral_features.png',
            dpi=150, bbox_inches='tight')
plt.show()
print('Saved fig5_spectral_features.png')
del ds_model; gc.collect(); torch.cuda.empty_cache()
```
